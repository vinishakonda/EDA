- now onwards we will write our code using try-except
- there are two blocks
- try block
- except block
- write your code in try block
- if any error occures in try block(original code) that code will catch in except

```python
a=10
b=20
c=a+b
print(f'the addition of {a} and {b} is:{c}')
```

```python
try:
    a=10
    b=20
    c=a+b
    print(f'the addition of {a} and {b} is:{c}')
except Exception as e:
    print(e)
```

```python
try:
    a=10
    print(a)
    b=20
    print(b)
    c=a/0
    print(f'the addition of {a} and {b} is:{c}')
except Exception as e:
    print(e)
    print("hahaha error milgaya")
```

```python
try:
    'a'+100
except Exception as e:
    print(e)
```

```python
try:
    import random
    random.random1()
except Exception as e:
    print(e)
```
