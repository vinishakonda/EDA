- before slash and after slash
- before: positional arguments
   -  Do not provide variable name
- after: keyword arguments
  - we can provide the variable name or no need to provide "our wish"

```python
def greet3(name,*,age):
    print(f"hello {name} and my age is: {age}")
```

```python
greet3('python',20)
# before * not there ====> correct
# after * not there ====> wrong
```

```python
greet3(name='python',20)
# before * there ====> correct
# after * not there ====> wrong
```

```python
greet3('python',age=20)
# before * not there ====> correct
# after * there ====> correct
```

```python
greet3(name='python',age=20)
# before * there ====> correct
# after * there ====> correct
```
